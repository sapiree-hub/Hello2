```python
#find the p-value for the skewness of sst2
from scipy.stats import skewtest
stat_Sst2, p_Sst2 = skewtest(Sst2)
Sst2

stat_Pvalb2, p_Pvalb2 = skewtest(Pvalb2)
Pvalb2

print(skewtest(Sst2).pvalue)
print(skewtest(Pvalb2).pvalue)
```

```python
#find the mean and median for Sst2 and Pvalb2
mean_Sst2 = np.mean(Sst2)
mean_Pvalb2 = np.mean(Pvalb2)
print("means Sst and Pvalb:")
print(mean_Sst2)
print(mean_Pvalb2)


median_Sst2 = np.median(Sst2)
median_Pvalb2 = np.median(Pvalb2)
print("medians Sst and Pvalb:")
print(median_Sst2)
print(median_Pvalb2)
```

```python
#Makes two subplots contained in one row (one row, two columns).

fig, ax = plt.subplots(1, 2)
```

```python
#Plots a histogram of Sst2
def plot_histograms(Sst2, Pvalb2):
  stat, p_Sst2 = skewtest(Sst2)
  fig, ax = plt.subplots(1, 2, figsize=(10,4))

  ax[0].hist(Sst2, bins=30)
  ax[0].set_xlabel("Sst2")
  ax[0].set_ylabel("Frequency")
  ax[0].set_title("Sst2 Hist.")
  if p_Sst2 > 0.05:
      ax[0].axvline(np.mean(Sst2), color='red',label='Mean')
  else:
      ax[0].axvline(np.median(Sst2), color='blue',label='Median')
  #Plots  for Pvalb2
  stat_Pvalb2, p_Pvalb2 = skewtest(Pvalb2)
  ax[1].hist(Pvalb2, bins=30)
  ax[1].set_xlabel("Pvalb2")
  ax[1].set_ylabel("Frequency")
  ax[1].set_title("Pvalb2 Hist.")
  if p_Pvalb2 > 0.05:
      ax[1].axvline(np.mean(Pvalb2), color='red')
  else:
      ax[1].axvline(np.median(Pvalb2), color='blue')

plot_histograms(Sst2, Pvalb2)
```
