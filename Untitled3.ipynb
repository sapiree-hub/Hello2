```python
# Here I import the required libraries
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
#this cell loads the CSV files from Github
v_rest_sst = np.loadtxt("https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv",delimiter=",")

v_rest_pvalb = np.loadtxt("https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv",delimiter=",")
```

```python
# This code shows the means of each data set
mean_sst = np.mean(v_rest_sst)
mean_pvalb = np.mean(v_rest_pvalb)

# This code performs the t-test
t_stat, p_value = stats.ttest_ind(v_rest_sst, v_rest_pvalb)

# then I make histograms only for p-value < 0.05
if p_value < 0.05:

    fig, ax = plt.subplots(1,2)

    ax[0].hist(v_rest_sst)
    ax[0].set_xlabel("Membrane Potential (mV)")
    ax[0].set_ylabel("Frequency")
    ax[0].set_title("SST")

    ax[1].hist(v_rest_pvalb)
    ax[1].set_xlabel("Membrane Potential (mV)")
    ax[1].set_ylabel("Frequency")
    ax[1].set_title("PVALB")

    plt.show()
```

```python
# This code defines which function to run the histograms
def vrestHist():

    mean_sst = np.mean(v_rest_sst)
    mean_pvalb = np.mean(v_rest_pvalb)

    t_stat, p_value = stats.ttest_ind(v_rest_sst, v_rest_pvalb)

    if p_value < 0.05:

        fig, ax = plt.subplots(1,2)

        ax[0].hist(v_rest_sst)
        ax[0].set_xlabel("Membrane Potential (mV)")
        ax[0].set_ylabel("Frequency")
        ax[0].set_title("SST")

        ax[1].hist(v_rest_pvalb)
        ax[1].set_xlabel("Membrane Potential (mV)")
        ax[1].set_ylabel("Frequency")
        ax[1].set_title("PVALB")

        plt.show()
```

```python
# show the histogram function
vrestHist()
```
