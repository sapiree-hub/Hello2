```python
# Importing required libraries
import numpy as np
import matplotlib.pyplot as plt
import statistics as st
```

```python
#loading the CSV files from Github
v_rest1 = np.loadtxt("https://raw.githubusercontent.com/hallockh/neur_265_spring2026/main/data/v_rest_sst.csv",delimiter=",")

v_rest2 = np.loadtxt("https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv",delimiter=",")
```

```python
# Stack arrays into a 2x13 matrix
v_rest = np.vstack((v_rest1, v_rest2))
```

```python
# Print shape of v_rest
print(v_rest.shape)
```

```python
# Return first row
v_rest[0]
```

```python
# Sum values in third column
v_rest[:, 2].sum()
```

```python
# Second-to-last value in second row
v_rest[1, -2]
```

```python
# Histogram of first row
plt.hist(v_rest[0])
plt.xlabel("Membrane Potential (mV)")
plt.ylabel("Frequency")
plt.title("Histogram of SST Resting Potentials")
plt.show()
```

```python
#this code cell should calculate the means
v_rest_mean_sst = st.mean(v_rest[0])
v_rest_mean_pvalb = st.mean(v_rest[1])

v_rest_mean_sst, v_rest_mean_pvalb
```

```python
#this code cell imports scpicy for t-tests
import scipy.stats as stats
```

```python
# Performing independent t-test
t_result = stats.ttest_ind(v_rest[0], v_rest[1])

t_statistic = t_result.statistic
p_value = t_result.pvalue

t_statistic, p_value
```
