```python
# Importing required libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Load Data
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv'
url3 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv'
url4 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv'

spikes_left_trial1 = np.loadtxt(url1, delimiter=",")
spikes_right_trial1 = np.loadtxt(url3, delimiter=",")
spikes_left_trial2 = np.loadtxt(url2, delimiter=",")
spikes_right_trial2 = np.loadtxt(url4, delimiter=",")
```

```python
# Print number of spikes in each trial
print("Left Trial 1:", len(spikes_left_trial1))
print("Right Trial 1:", len(spikes_right_trial1))
print("Left Trial 2:", len(spikes_left_trial2))
print("Right Trial 2:", len(spikes_right_trial2))
```

```python
# Compute firing rates (spikes per second)
duration = 30  # seconds

fr_left_t1 = len(spikes_left_trial1) / duration
fr_right_t1 = len(spikes_right_trial1) / duration
fr_left_t2 = len(spikes_left_trial2) / duration
fr_right_t2 = len(spikes_right_trial2) / duration

print("Firing Rates (Hz):")
print("Trial 1 Left:", fr_left_t1)
print("Trial 1 Right:", fr_right_t1)
print("Trial 2 Left:", fr_left_t2)
print("Trial 2 Right:", fr_right_t2)
```

```python
# Plot spike trains for trial 1 and trial 2

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Trial 1
axes[0].scatter(spikes_left_trial1, np.ones_like(spikes_left_trial1), color='red', label='Left')
axes[0].scatter(spikes_right_trial1, np.ones_like(spikes_right_trial1)*2, color='blue', label='Right')
axes[0].set_title("Trial 1 Spike Train")
axes[0].set_xlabel("Time (s)")
axes[0].set_yticks([1,2])
axes[0].set_yticklabels(["Left", "Right"])
axes[0].legend()

# Trial 2
axes[1].scatter(spikes_left_trial2, np.ones_like(spikes_left_trial2), color='red', label='Left')
axes[1].scatter(spikes_right_trial2, np.ones_like(spikes_right_trial2)*2, color='blue', label='Right')
axes[1].set_title("Trial 2 Spike Train")
axes[1].set_xlabel("Time (s)")
axes[1].set_yticks([1,2])
axes[1].set_yticklabels(["Left", "Right"])
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# Inter-spike intervals (ISI)
isi_left_t1 = np.diff(spikes_left_trial1)
isi_right_t1 = np.diff(spikes_right_trial1)
isi_left_t2 = np.diff(spikes_left_trial2)
isi_right_t2 = np.diff(spikes_right_trial2)
```

```python
# Histogram bins
bins = np.arange(0, 2, 0.01)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0,0].hist(isi_left_t1, bins=bins)
axes[0,0].set_title("ISI Left Trial 1")

axes[0,1].hist(isi_right_t1, bins=bins)
axes[0,1].set_title("ISI Right Trial 1")

axes[1,0].hist(isi_left_t2, bins=bins)
axes[1,0].set_title("ISI Left Trial 2")

axes[1,1].hist(isi_right_t2, bins=bins)
axes[1,1].set_title("ISI Right Trial 2")

for ax in axes.flat:
    ax.set_xlabel("ISI (s)")
    ax.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
# Time bins
time_bins = np.arange(0, 30, 1)

def compute_fr_over_time(spikes):
    counts, _ = np.histogram(spikes, bins=np.arange(0, 31, 1))
    return counts  # spikes per 1-second bin

fr_left_t1_time = compute_fr_over_time(spikes_left_trial1)
fr_right_t1_time = compute_fr_over_time(spikes_right_trial1)
fr_left_t2_time = compute_fr_over_time(spikes_left_trial2)
fr_right_t2_time = compute_fr_over_time(spikes_right_trial2)

fig, axes = plt.subplots(1, 2, figsize=(12,4))

# Trial 1
axes[0].plot(time_bins, fr_left_t1_time, color='red', label='Left')
axes[0].plot(time_bins, fr_right_t1_time, color='blue', label='Right')
axes[0].set_title("Trial 1 Firing Rate Over Time")
axes[0].set_xlabel("Time (s)")
axes[0].set_ylabel("Spikes / s")
axes[0].legend()

# Trial 2
axes[1].plot(time_bins, fr_left_t2_time, color='red', label='Left')
axes[1].plot(time_bins, fr_right_t2_time, color='blue', label='Right')
axes[1].set_title("Trial 2 Firing Rate Over Time")
axes[1].set_xlabel("Time (s)")
axes[1].set_ylabel("Spikes / s")
axes[1].legend()

plt.tight_layout()
plt.show()
```
